**Validación cruzada (K-Fold):** técnica que divide los datos en *k* partes o "bolsas" del mismo tamaño. En cada iteración, una bolsa se usa como test y las demás como entrenamiento, hasta que todas las bolsas han sido test una vez. Así cada dato se usa tanto para entrenar como para evaluar, y el resultado es más confiable que con una sola división..

### Importación de librerías
* **KFold**: clase de scikit-learn que divide los datos en *k* bolsas para hacer validación cruzada.

```python
from sklearn.model_selection import KFold
```

### Carga de datos
Creamos una lista sencilla de 10 elementos, de la "a" a la "j", para ver con claridad cómo se reparten los datos en cada bolsa.

```python
# Carga de datos.
X = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"]
```

### Configuración de la validación cruzada
* `n_splits=5` divide los datos en **5 bolsas**. Como tenemos 10 elementos, cada bolsa tiene 2.
* `shuffle=True` mezcla los datos antes de dividirlos, para que las bolsas no se formen en orden.
* `split()` genera las divisiones de entrenamiento y test de cada iteración.

```python
# Validación cruzada.
kf = KFold(n_splits = 5, shuffle=True) # Shuffle permite aleatorizar las bolsas de test vez de hacer una lectura por orden 
bolsas = kf.split(X)
print(bolsas)
```

**Interpretación:** al imprimir `bolsas` no aparecen los datos, sino un **generador** (`generator object`). Esto significa que las divisiones no se calculan de golpe, sino una por una cuando las recorremos con un ciclo `for`.

### Mostrar cada iteración
Recorremos las bolsas con un ciclo `for`. En cada iteración mostramos qué posiciones se usan para entrenamiento y cuáles para test, y con `k` llevamos la cuenta del número de iteración.

```python
# Mostrar ejemplos de cada iteración.
k = 1
for train, test in bolsas:
    print("Iteracion", k, ":")
    print(" - Entrenamiento: %s" % (train)) 
    print(" - Test: %s" % (test))
    k = k + 1
```

**Interpretación:** se realizan 5 iteraciones. En cada una, 8 elementos se usan para entrenamiento y 2 para test. Los valores que aparecen son las **posiciones** (índices) de los datos, no las letras: por ejemplo, el 0 es "a" y el 9 es "j". Todos los elementos aparecen en test **una sola vez**, que es la idea principal de la validación cruzada.
