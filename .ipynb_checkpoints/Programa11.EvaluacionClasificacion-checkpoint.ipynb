**Validación cruzada estratificada y métricas de clasificación:** primero separamos un conjunto de test con *hold out* (80-20%). Luego, dentro del entrenamiento, aplicamos validación cruzada 

**Stratified K-Fold**, que mantiene en cada bolsa la misma proporción de clases que en los datos originales. Con `cross_val_predict` obtenemos una predicción para cada instancia del entrenamiento y con ellas calculamos exactitud, precisión, sensibilidad, F1-score y la matriz de confusión. Al final entrenamos el modelo definitivo y lo evaluamos con el test.

### Importación de librerías
* **numpy**: manejo de arreglos y dimensiones de los datos.
* **datasets**: trae el dataset de ejemplo Iris.
* **cross_val_predict**: devuelve la predicción de cada instancia cuando le toca estar en la bolsa de validación.
* **cross_validate / cross_val_score**: calculan métricas en cada bolsa de la validación cruzada.
* **KFold / StratifiedKFold**: dividen los datos en *k* bolsas. La versión estratificada respeta la proporción de cada clase.
* **StandardScaler / preprocessing**: estandarizan los atributos (media 0 y desviación 1).
* **metrics**: funciones para evaluar el modelo (exactitud, precisión, matriz de confusión, etc.).
* **train_test_split**: separa los datos en entrenamiento y test.
* **Pipeline**: une varios pasos (estandarizar y clasificar) en un solo modelo.
* **pprint**: imprime los resultados de forma ordenada.
* **clone**: crea una copia limpia del modelo, sin entrenar.
* **SVC**: clasificador de máquinas de vectores de soporte.

### Carga de datos
Cargamos el dataset **Iris** y revisamos sus dimensiones.

```python
# Carga de datos.
datos = datasets.load_iris()
print(np.shape(datos.data))
```

**Interpretación:** el dataset tiene 150 instancias y 4 atributos.

### Separar atributos y variable objetivo
Guardamos los atributos en `X` y la clase a predecir en `y`.

```python
# Extraer la matriz de datos "X" y la variable target "y"
X = datos.data
y = datos.target
```

### Partición externa: hold out 80-20%
Separamos el 20% de los datos para test y el 80% para entrenamiento. El test se reserva para la evaluación final.

```python
# Partición EXTERNA. Test: hold-out split 80-20%.
X_train, X_test, y_train, y_test = train_test_split(datos.data, datos.target, test_size=0.2, random_state=42)
print('Train dimensions: ', np.shape(X_train))
print('Test dimensions:  ', np.shape(X_test))
```

### Estandarizador
Creamos el `StandardScaler`. No lo aplicamos directamente a los datos, sino que lo incluimos dentro del pipeline para evitar la fuga de datos.

```python
# Estandarización de los datos de entrenamiento y test
#standardizer = preprocessing.StandardScaler()
#X_train = standardizer.fit_transform(X_train)
#X_test = standardizer.transform(X_test)
standardizer = StandardScaler()
```

### Definición del modelo con Pipeline
Unimos en un **Pipeline** la estandarización y el modelo **SVC**.
* `C=0.1`: valor bajo de regularización. El modelo es más simple y tolera algunos errores.
* `kernel='rbf'`: permite separar clases con fronteras no lineales.
* `gamma='auto'`: define la influencia de cada punto como 1 dividido para el número de atributos.

```python
alg = Pipeline([
    ("scaler", standardizer ),
    ("svc", SVC(C=0.1,gamma="auto", kernel="rbf"))
])
```

### Validación cruzada interna estratificada
* `StratifiedKFold` divide el entrenamiento en **5 bolsas** con la misma proporción de las 3 clases. `shuffle=True` mezcla los datos y `random_state=42` hace que el resultado sea repetible.
* `cross_val_predict` entrena el modelo con 4 bolsas y predice la bolsa restante, hasta pasar por las 5. Así obtenemos **una predicción para cada una de las 120 instancias** del entrenamiento.
* Mostramos las predicciones (`y_pred`) junto a las clases reales (`y_train`) para compararlas.

```python
# Definimos el algoritmo
alg = SVC(C=0.1, gamma='auto', kernel='rbf')
```

**Interpretación:** los dos arreglos son casi iguales. Las diferencias son los errores del modelo, que se ven con más detalle en la matriz de confusión.

```python
# Cross-validation interno
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
# Utilizar comando "cross_val_predict" para extraer directamente las predicciones de las 5 bolsas de validación
y_pred = cross_val_predict(alg, X_train, y_train, cv=cv)
pprint(y_pred)
pprint(y_train)
```

```python
# Presentación de los resultados de la evaluación.
print("Exactitud: %.3f" % (metrics.accuracy_score(y_train, y_pred))) # accuracy
print("Precisión: %.3f" % (metrics.precision_score(y_train, y_pred, average="macro"))) # precision
print("Sensibilidad: %.3f" % (metrics.recall_score(y_train, y_pred, average="macro"))) # sensibilidad
print("F1-score: %.3f" % (metrics.f1_score(y_train, y_pred, average="macro"))) # F-score
```

```python
# Extraemos la matriz de confusión
print("Matriz de confusión:\n", metrics.confusion_matrix(y_train, y_pred))
```

```python
print("Tabla de métricas:\n", metrics.classification_report(y_train, y_pred))
```

```python
# Una vez hemos optimizado los hiperparámetros con el conjunto de validación, utilizamos todos los datos de
# "train" y "validation" para entrenar un modelo definitivo con esos hiperparámetros ajustados
model = clone(alg) # Definimos el modelo con los hiperparámetros óptimos
model = model.fit(X_train, y_train)# Entrenamos el modelo con todos los datos de "train" + "val"

# y finalmente evaluamos el test con el método "score()"
test_results = model.score(X_test, y_test)
print('Exactitud en test: ', np.round(test_results*100, 4), '%')
```

```python
# obtenemos las predicciones del test para extraer el resto de métricas de interés
y_pred_test = model.predict(X_test) # Extraer predicciones
print("Tabla de metricas de test:\n", metrics.classification_report(y_test, y_pred_test)) # Aplicar el método de classification_
print("Matriz de confusión:\n", metrics.confusion_matrix(y_test, y_pred_test)) # Extraer la matriz de confusión
```
