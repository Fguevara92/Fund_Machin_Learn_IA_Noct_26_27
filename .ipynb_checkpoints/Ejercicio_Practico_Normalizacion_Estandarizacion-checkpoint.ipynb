### Antecedentes/Fundamentación teórica.

En el contexto del aprendizaje automático, los algoritmos suelen verse afectados por la escala de los datos. Variables con valores grandes pueden dominar el comportamiento del modelo, generando sesgos y resultados pocos conficables. Para mitigar este problema, se emplea técnicas como:

* **Normalización:** transforma los datos a un rango especifico (generalmente[0,1])
* **Estandarización:** ajusta los datos para que tengan media 0 y desviación estandar 1. Estas técnicas son fundamentales en algoritmos como:
  * KNN
  * Regresión logística
  * Redes neuronales

**X** son las características, es decir las medidas que el modelo va a usar para aprender.

**Análisis exploratorio**

Resume cada columna con el promedio, la desviación, el mínimo, el máximo y los cuartiles.

```python
#Exploramos o generamos un analis exploratorio del conjutno de datos
X.describe()
```

### Conclusión.

Cuando observamso la tabla, lo primero que se observa es que cada variable tiene valores diferentes en tamaño y rango Por ejemplo:

* **PetalLengthCm** tiene valores desde 1.0 hasta 6.9, lo que significa que varia bastante
* En cambio, **SepalWidthCm** va aproximadamente de 2.0 a 4.4 es decir, tiene meno variación. Esto nos dice que no todas las variables están en la misma escala.

### Qué significa?

Al ver los datos vemos que algunas variables tiene números más grandes que otras, y en modelos como KNN.

* Los números grandes “pesan más”
* Influyen mas en el resultado. Sino hacemos nada y trabajamos con estos datos el mnodelo va a tomas decisiones basadas en el argumeto **PetalLengtnCm** y menos en otras variables.Esto no es lo conveniento y a esto lo llamamos sesgo. Para no tener sesgo dentro de nuestros datos lo que hacemso es:
  * Normalización.
  * Estandarización

Esto permite que las variables esten en la misma celda y tengan la misma importancia.

**Gráfico de datos originales**



```python
#dibujamos los datos originales
sns.boxplot(data=X)
#Ponemos un titulo a la imagen
plt.title("Datos originales")
plt.show()
```

## Interpretación de la imagen

Este gráfico nos permite observar que los datos no están en la misma escala y que algunas variables
tienen mayor variabilidad que otras. Esto puede afectar el comportamiento de los modeos de
inteligencia artifical, por lo que es necesario aplicar técnicas como la normalización o estandarización
para equilibrar la influencia de todas las varaibles.

**Normalización (Min-Max)**

Creamos el normalizador y aprende el mínimo y el máximo de cada columna, y transform lleva todos los valores al rango de 0 a 1. Y el resultado sale como arreglo sin nombres, así que lo volvemos a convertir en tabla con sus columnas, al finalizar comprobamos que ahora los valores están entre 0 y 1.

```python
#Normalizamos el conjunto de datos
normalizer = preprocessing.MinMaxScaler()
X_norm=normalizer.fit_transform(X)
X_norm = pd.DataFrame(X_norm, columns=X.columns)
X_norm.head()
```

```python
X_norm.describe()
```

**Estandarización**

Creamos el estandarizador, a cada valor le resta el promedio y lo divide para la desviación. Así cada columna queda con media 0 y desviación 1

```python
#Estandarización
standarizer = preprocessing.StandardScaler()
#Estandarizamos los datos
X_std=standarizer.fit_transform(X)
X_std= pd.DataFrame(X_std, columns=X.columns)
X_std.head()
```

**Comparación visual**

Crea un lienzo ancho para poner 3 gráficos.

```python
#Comporación visual
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
sns.boxplot(data=X)
plt.title("Original")
plt.subplot(1,3,2)
sns.boxplot(data=X_norm)
plt.title("Normalizado")
plt.subplot(1,3,3)
sns.boxplot(data=X_std)
plt.title("Estandarizado")
plt.show()
```

**Interpretación**

* En el gráfico Original las cajas están a distintas alturas. 
* En Normalizado todas quedan entre 0 y 1.
* En Estandarizado todas se centran alrededor de 0, más o menos entre -2 y 3.

La forma de cada caja no cambia: los datossiguen diciendo lo mismo, solo cambiamos la “regla” con la que los medimos.

**Conclusión**

Los datos de Iris estaban completos y limpios, pero cada variable venía en una escala distinta. Esto puede hacer que modelos como KNN, regresión logística o redes neuronales le den más importancia a las variables con números grandes. Con la normalización llevamos todo al rango de 0 a 1, y con la estandarización centramos todo en 0 con la misma dispersión. 

En los dos casos las variables quedan comparables y con el mismo peso, sin perder la información original.
