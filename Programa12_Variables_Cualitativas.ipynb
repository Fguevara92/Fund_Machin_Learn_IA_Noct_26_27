**Variables cualitativas:** son las que describen una cualidad o categoría en lugar de un número, como el nivel de estudios de una persona. Para analizarlas contamos cuántas veces se repite cada categoría (frecuencias) y las representamos con gráficos como el diagrama de barras y el de sectores. En este ejercicio el nivel de estudios es además **ordinal**, porque sus categorías tienen un orden natural.

### Importación de librerías
* **numpy**: manejo de arreglos y conteo de categorías.
* **matplotlib.pyplot**: creación de gráficos.
* **random**: generación de valores al azar. Con `random.seed(42)` fijamos la semilla.

```python
# Importamos Librerias para la creacion de datos
import numpy as np
import matplotlib.pyplot as plt
import random
random.seed(42)
```

### Creación de la estructura de datos
* Creamos una lista vacía y con `extend()` agregamos cada nivel de estudios tantas veces como personas lo tienen: 110 de Secundaria, 172 de Bachillerato, 145 de Grado, 52 de Master y 21 de Doctorado.
* Con `np.random.permutation()` mezclamos los datos para que no queden ordenados, como pasaría en una encuesta real.

```python
# Creo la estructura de datos
nivel_estudios = []
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)
nivel_estudios = np.random.permutation(nivel_estudios)
```

### Visualización del conjunto de datos
Imprimimos los 500 datos para comprobar que la lista se creó bien y que los niveles quedaron mezclados.

```python
# imprimimos el conjunto de datos
print(nivel_estudios)
```

### Frecuencias con NumPy
Con `np.unique()` y `return_counts=True` obtenemos las categorías sin repetir y cuántas veces aparece cada una, es decir, su **frecuencia absoluta**.

```python
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

**Interpretación:** Bachillerato es el nivel más frecuente con 172 personas y Doctorado el menos frecuente con 21. NumPy ordena las categorías **alfabéticamente**, no por nivel de estudios.

### Diagrama de barras
Cada barra representa un nivel de estudios y su altura indica cuántas personas lo tienen. Es el gráfico más usado para comparar categorías de una variable cualitativa.

```python
# realizamos un diagrama de barras para entender como estan los datos

plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras de estudios en la ciudad de Valencia')
plt.bar(estudios, frecuencia)
plt.show()
```

**Interpretación:** la barra más alta corresponde a Bachillerato y la más baja a Doctorado. Las barras salen en orden alfabético, por eso no siguen el orden lógico de los niveles.

### Diagrama de sectores
Cada parte del círculo representa la proporción de un nivel de estudios dentro del total. Con `autopct='%1.2f%%'` mostramos el porcentaje con dos decimales.

```python
# Dibujamos un diagrama de sectores o pie

colores = {'yellow', 'red', 'green', 'blue', 'magenta'}
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```

**Interpretación:** Bachillerato ocupa la mayor parte del círculo (34,40%) y Doctorado la menor (4,20%). Los colores están en un conjunto `{ }`, que no guarda orden, por eso pueden cambiar de nivel en cada ejecución.

## Tarea: implementación con pandas
### Importación de pandas
* **pandas**: librería para trabajar con datos en forma de tablas, parecido a una hoja de Excel pero desde código.

```python
# Tarea implementar con pandas esta información
import pandas as pd
```

### Creación del DataFrame
Convertimos el arreglo de NumPy en un **DataFrame**, que es una tabla con filas y columnas. Con `head(10)` mostramos las primeras 10 filas.

```python
df = pd.DataFrame({'nivel_estudios': nivel_estudios})
df.head(10)
```

### Resumen de los datos
Mostramos el total de registros y con `describe()` obtenemos un resumen rápido de la variable.

```python
print('Total de registros:', len(df))
df.describe()
```

**Interpretación:** hay 500 registros y 5 categorías distintas (`unique`). La **moda** (`top`), es decir, el valor que más se repite, es Bachillerato con 172 personas (`freq`).

### Frecuencias con pandas
Con `value_counts()` contamos cuántas veces aparece cada nivel. Hace lo mismo que `np.unique()`, pero en una sola línea y ordenado de mayor a menor.

```python
frecuencia_pd = df['nivel_estudios'].value_counts()
frecuencia_pd
```

### Ordenar los niveles
Con `reindex()` acomodamos las frecuencias en el orden lógico, de Secundaria a Doctorado. Como la variable es **ordinal**, así la tabla y los gráficos se leen mejor.

```python
orden = ['Secundaria', 'Bachillerato', 'Grado', 'Master', 'Doctorado']
frecuencia_pd = frecuencia_pd.reindex(orden)
frecuencia_pd
```

### Tabla de frecuencias
* **Frecuencia absoluta (fi):** número de personas en cada nivel.
* **Frecuencia relativa (hi):** frecuencia absoluta dividida para el total de datos.
* **Porcentaje (%):** frecuencia relativa multiplicada por 100.
* **Frecuencia absoluta acumulada (Fi):** suma de las frecuencias hasta ese nivel. Se calcula con `cumsum()`.
* **Frecuencia relativa acumulada (Hi):** lo mismo, pero en proporción.

```python
tabla = pd.DataFrame({'Frecuencia absoluta (fi)': frecuencia_pd})
tabla['Frecuencia relativa (hi)'] = tabla['Frecuencia absoluta (fi)'] / tabla['Frecuencia absoluta (fi)'].sum()
tabla['Porcentaje (%)'] = tabla['Frecuencia relativa (hi)'] * 100
tabla['Frec. absoluta acumulada (Fi)'] = tabla['Frecuencia absoluta (fi)'].cumsum()
tabla['Frec. relativa acumulada (Hi)'] = tabla['Frecuencia relativa (hi)'].cumsum()
tabla.index.name = 'Nivel de estudios'
tabla.round(3)
```

**Interpretación:** el 34,4% de las personas tiene Bachillerato y solo el 4,2% tiene Doctorado. En la acumulada vemos que 282 personas (56,4%) tienen como máximo Bachillerato, y la última fila llega a 500 y a 1, que es el total.

### Diagrama de barras con pandas
Con `.plot(kind='bar')` hacemos el gráfico directamente desde los datos de pandas. Ahora las barras salen en el orden lógico de los niveles.

```python
ax = frecuencia_pd.plot(kind='bar', color='steelblue', edgecolor='black', rot=0)
ax.set_title('Diagrama de barras de estudios en la ciudad de Valencia')
ax.set_xlabel('Nivel de estudios')
ax.set_ylabel('Frecuencia absoluta')
plt.show()
```

**Interpretación:** con las barras ordenadas se ve que la mayoría de personas se concentra en Bachillerato y Grado, y que la cantidad baja mucho en Master y Doctorado.

### Diagrama de sectores con pandas
Con `.plot(kind='pie')` hacemos el gráfico de pastel. Usamos una **lista** `[ ]` de colores para que cada nivel tenga siempre el mismo color, y con `set_ylabel('')` quitamos la etiqueta que pandas pone por defecto.

```python
colores_lista = ['yellow', 'red', 'green', 'blue', 'magenta']

ax = frecuencia_pd.plot(kind='pie', colors=colores_lista, autopct='%1.2f%%', startangle=90)
ax.set_title('Diagrama de sectores')
ax.set_ylabel('')
plt.show()
```

**Interpretación:** Bachillerato (34,40%) y Grado (29,00%) suman más de la mitad de los datos, mientras que Master (10,40%) y Doctorado (4,20%) son los grupos más pequeños.
