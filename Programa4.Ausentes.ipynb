**Imputación de valores ausentes:** proceso en el cual reemplazamos los datos faltantes (`NaN`) por un valor estimado, como la media, la mediana o el valor más frecuente de la columna, para que el conjunto de datos quede completo y pueda usarse en un modelo.

**Link Repositorio** https://github.com/Fguevara92/Fund_Machin_Learn_IA_Noct_26_27.git


### Importación de librerías
* **numpy**: nos permite crear los arreglos de datos y representar los valores ausentes con `np.nan` (*Not a Number*).
* **SimpleImputer**: herramienta de scikit-learn que reemplaza los valores ausentes por un valor calculado a partir de los datos, como la media, la mediana o la moda.

```python
import numpy as np
from sklearn.impute import SimpleImputer
```

### Carga de datos de entrenamiento
Creamos un pequeño conjunto de entrenamiento `X_train` con 3 filas y 2 columnas. La segunda fila tiene un valor ausente (`np.nan`) en la primera columna. Este es el tipo de dato faltante que debemos completar antes de entrenar un modelo, porque la mayoría de algoritmos no aceptan valores vacíos.

```python
# Carga de datos
X_train = [[1, 4],
           [np.nan, 3],
           [7, 8]]
```

### Imputación de valores ausentes en el entrenamiento
* Creamos el imputador con `strategy='mean'`, es decir, cada valor ausente se reemplaza por la **media** de su columna.
* `fit()` **aprende** la media de cada columna usando solo los valores conocidos.
* `transform()` **reemplaza** los `NaN` por esas medias.
* `fit_transform()` hace los dos pasos en una sola línea.

```python
# Limpieza de datos: imputación valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizamos la estrategia de la media
# Aplicar los métodos "fit" y "transform" para imputar los valores pérdidos de X_train
X_train_clean = imputer.fit_transform(X_train)
print(X_train_clean)
```

**Interpretación:** el imputador calculó la media de cada columna con los valores conocidos:
* Columna 1: (1 + 7) / 2 = **4** (el `NaN` no se cuenta).
* Columna 2: (4 + 3 + 8) / 3 = **5**.

El valor ausente de la segunda fila se reemplazó por **4**, la media de la columna 1. El resultado es `[[1, 4], [4, 3], [7, 8]]`, un conjunto completo y sin valores faltantes.

### Imputación de valores ausentes en el conjunto de test
Creamos un conjunto de prueba `X_test` con dos valores ausentes: uno en la primera columna y otro en la segunda.

Aquí usamos solo `transform()`, **no** `fit()`. Así el imputador reemplaza los `NaN` con las medias que **aprendió del entrenamiento**, no con las medias del test. Esto es importante para evitar la **fuga de datos** (*data leakage*): el conjunto de test representa datos nuevos que el modelo no debe "conocer" de antemano.

```python
# Imputación de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
          [6, np.nan],
          [8, 2]]

# Imputar los datos del conjunto de test
X_test_clean=imputer.transform(X_test)
print(X_test_clean)
```

**Interpretación:**
* El `NaN` de la primera columna se reemplazó por **4**.
* El `NaN` de la segunda columna se reemplazó por **5**.

Son las medias del conjunto de **entrenamiento**. Si hubiéramos calculado las medias del test, los valores habrían sido distintos: (6 + 8) / 2 = 7 en la primera columna y (10 + 2) / 2 = 6 en la segunda. Esto demuestra que `transform()` aplica lo aprendido en el entrenamiento y no recalcula con los datos de prueba.
